# Machine Learning
[*Machine learning*](https://www.geeksforgeeks.org/machine-learning/machine-learning/) is a subset of Artificial Intelligence that focuses on developing models and algorithms that let computers learn from data without being explicitly programmed for every task. 

</br>

## Why Machine Learning?
Traditionally, a developer needed to tediously code every fine detail in order for a computer to perform a task. This, however, is impossible to do for real-world data, which is unorganized and complex.

Instead, a different approach is to give the computer large datasets and statistical algorithms so that it can analyze the data, find patterns, and extrapolate meaning from them. Furthermore, the computer can improve its own accuracy over time. This is called Machine Learning, where the machine can learn by itself without humans needing to hardcode every feature.

Hence, for real-world problems such as image processing and language processing where coding every rule isn't viable, we use **Machine Learning**.

### Supervised Learning:

Model is trained using [*labelled data*](https://www.geeksforgeeks.org/machine-learning/what-is-the-difference-between-labeled-and-unlabeled-data/). This type of model learns the realationship between input and labelled("correct data") and is used for predictions.

</br>

### Unsupervised Learning:

Model is trained using [*unlabeled data*](https://www.geeksforgeeks.org/machine-learning/what-is-the-difference-between-labeled-and-unlabeled-data/). This type of model finds patterns in unlabelled data. This type of model is used to arrange the data in meaningfull clusters or reduce dimensions of the data.

</br>

### Reinforcement Learning: 

Model learns through trial and error.
